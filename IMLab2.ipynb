Критерий серий
Составим вариационный ряд, т. е. расположим хi в порядке возрастания x1, x2, ... xN. Вычислим выборочное значение медианы по формулам:


```python
varline = sorted(ui)
if len(ui)%2 == 0:
    median = 0.5*(varline[int(len(ui)/2-1)]+varline[int((len(ui))/2)])
else:
    median = varline[len(ui)/2]
    
print("Медиана ",median)
```

Возвращаемся к исходной последовательности оценок x1, x2, ... xN. Сравнивая каждое хi с  , ставим знак «+», если хi ≥   и знак «-», если хi < . Образуется последовательность вида: +++ - ++ ------ +. Серией называется последовательность однотипных наблюдений, после которой следуют наблюдения противоположного типа или же вообще нет никаких наблюдений. 

```python
plmn = []
for i in range(len(ui)):
    if ui[i] < median:
        plmn.append("-")
    else:
        plmn.append("+")

print(plmn)
```

```python
counter = 1
previous = plmn[0]
for i in range(len(plmn)):
    if(plmn[i] != previous):
        counter += 1
        previous = plmn[i]
print(len(plmn))
print("Crit value up ",61)
print("Check value ",counter)
print("Crit value down ",40)
```
